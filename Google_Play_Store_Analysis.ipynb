```python
apps.drop_duplicates(inplace=True)

print("Duplicates removed successfully")
print(apps.shape)
```

```python
apps[~apps['Installs'].str.contains(r'\d', na=False)]['Installs'].unique()
```

```python
apps = apps[apps['Installs'] != 'Free']

print("Removed incorrect Installs rows")
print(apps.shape)
```

```python
apps['Installs'] = (
    apps['Installs']
    .str.replace("+", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(int)
)

apps[['App','Installs']].head()
```

```python
apps['Installs'].dtype
```

```python
apps['Price'].unique()[:20]
```

```python
apps['Price'] = (
    apps['Price']
    .str.replace("$", "", regex=False)
    .astype(float)
)

apps[['App','Price']].head()
```

```python
apps['Price'].dtype
```

```python
apps['Size'].unique()[:20]
```

```python
def convert_size(size):
    
    if isinstance(size, str):
        
        if "M" in size:
            return float(size.replace("M", ""))
        
        elif "k" in size:
            return float(size.replace("k", "")) / 1024
        
        elif "Varies with device" in size:
            return np.nan
    
    return np.nan


apps['Size_MB'] = apps['Size'].apply(convert_size)

apps[['Size', 'Size_MB']].head()
```

```python
apps[['Size','Size_MB']].sample(10)
```

```python
apps['Size_MB'].isnull().sum()
```

```python
apps['Size_MB'].fillna(
    apps['Size_MB'].median(),
    inplace=True
)

print("Missing Size values handled")
```

```python
apps['Rating'].isnull().sum()
```

```python
apps['Rating'].fillna(
    apps['Rating'].mean(),
    inplace=True
)

print("Rating missing values handled")
```

```python
category_count = apps['Category'].value_counts()

plt.figure(figsize=(12,6))

sns.barplot(
    x=category_count.values,
    y=category_count.index
)

plt.title("Distribution of Apps Across Categories")
plt.xlabel("Number of Apps")
plt.ylabel("Category")

plt.show()
```

```python
reviews.dropna(
    subset=['Translated_Review'],
    inplace=True
)

reviews.shape
```

```python
from textblob import TextBlob

def get_sentiment(text):

    polarity = TextBlob(text).sentiment.polarity

    if polarity > 0:
        return "Positive"

    elif polarity < 0:
        return "Negative"

    else:
        return "Neutral"


reviews['New_Sentiment'] = (
    reviews['Translated_Review']
    .apply(get_sentiment)
)

reviews.head()
```

```python
plt.figure(figsize=(7,4))

sns.countplot(
    data=reviews,
    x='New_Sentiment'
)

plt.title("User Review Sentiment Analysis")

plt.show()
```

```python
sentiment_data = pd.merge(
    reviews,
    apps[['App','Category']],
    on='App',
    how='inner'
)

sentiment_data.head()
```

```python
sentiment_category = (
    sentiment_data
    .groupby(['Category','New_Sentiment'])
    .size()
    .reset_index(name='Count')
)


plt.figure(figsize=(12,6))

sns.barplot(
    data=sentiment_category,
    x='Category',
    y='Count',
    hue='New_Sentiment'
)

plt.xticks(rotation=90)

plt.title(
    "Sentiment Analysis by App Category"
)

plt.show()
```
